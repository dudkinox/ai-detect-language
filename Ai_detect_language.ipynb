### วิเคราะห์ข้อมูล

```python
def detectLanguagesModelAI(text, charMap):
    detectedLanguages = set()
    cleanedText = text.replace(" ", "")
    for chars, lang in charMap.items():
        if any(char in cleanedText for char in chars):
            detectedLanguages.add(lang)
    return list(detectedLanguages)
```

```python
charMap = {row['chars']: row['value'] for _, row in df.iterrows()}
```

```python
def displayLanguage(detectedLangs):
  if len(detectedLangs) == 0:
    print("404 Not found")
    return
  for lang in detectedLangs:
    print(f"- {lang}")
```

```python
inputText = input(str("Enter word => "))
detectedLangs = detectLanguagesModelAI(inputText, charMap)
print("\n=== Detected Languages ===")
displayLanguage(detectedLangs)
print("=========================")
```
